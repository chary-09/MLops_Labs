```python
# Step 11: Correctly Commit Version 1 and then Version 2

# Ensure we are in the correct DVC project directory
%cd /content/DVC_Project

# Clean up existing git repository and dvc files to start fresh
!rm -rf .git
!rm -rf .dvc
!rm -f student.csv.dvc

# Initialize Git and DVC again
!git init
!dvc init

# Set Git user configuration (if not already set globally or for this repo)
# These should ideally be set once, but re-setting for robustness after reset
!git config user.name "chary"
!git config user.email "charybhanuprasad07@gmail.com"

# --- Create and Commit Dataset Version 1 ---
# Create 'Dataset Version 1' (original data without 'Sita')
import pandas as pd
data_v1 = {
    "Student": ["Ravi", "Priya", "Anu", "Rahul"],
    "Marks": [90, 85, 78, 95]
}
df_v1 = pd.DataFrame(data_v1)
df_v1.to_csv("student.csv", index=False)

# Add 'Dataset Version 1' to DVC
!dvc add student.csv

# Add DVC-related files and the dvc-tracked file to Git
!git add .gitignore .dvcignore .dvc/config student.csv.dvc

# Commit Dataset Version 1
!git commit -m "Dataset Version 1"

# --- Create and Commit Dataset Version 2 ---
# Create 'Dataset Version 2' (adding 'Sita')
data_v2 = {
    "Student": ["Ravi", "Priya", "Anu", "Rahul", "Sita"],
    "Marks": [90, 85, 78, 95, 88]
}
df_v2 = pd.DataFrame(data_v2)
df_v2.to_csv("student.csv", index=False)

# Add the modified 'student.csv' (version 2) to DVC
!dvc add student.csv

# Add the updated student.csv.dvc file to Git
!git add student.csv.dvc

# Commit Dataset Version 2
!git commit -m "Dataset Version 2"

# Display the corrected Git version history
!git log --oneline
```
